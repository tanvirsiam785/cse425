# Demo: end-to-end musical context understanding

Loads the best Task 3 model (cross-attention GNN-BERT fusion, Macro-F1 0.611) and the
Task 4 dual-encoder to demonstrate:
1. Predicting tags for a given (caption, audio) pair.
2. Retrieving the closest audio clips for a free-text query caption.

Requires trained checkpoints (see `src/train.py`) and processed data under `data/processed/`.

```python
import sys, pathlib, torch
sys.path.append(str(pathlib.Path('..').resolve() / 'src'))

from fusion_model import GNNBertFusionModel
from contrastive import DualEncoder
from bert_encoder import BertTagEncoder

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```
